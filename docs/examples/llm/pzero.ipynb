# PZERO

[PZERO](https://pzero.studio) is a prepaid inference marketplace in front of Venice-only upstream that exposes an OpenAI-compatible API.

You can use it directly with LlamaIndex using the `OpenAILike` integration class by pointing `api_base` to `https://api.pzero.studio/v1` and setting `is_chat_model=True`.

> **Notes:**
> - API keys can be acquired at [pzero.studio/agents](https://pzero.studio/agents).
> - Available text model IDs can be retrieved via `GET https://api.pzero.studio/v1/models`.

```python
import os

pzero_api_key = os.environ.get("PZERO_API_KEY", "your-pzero-api-key")
```

```python
from llama_index.llms.openai_like import OpenAILike

llm = OpenAILike(
    model="deepseek-v4-flash",
    api_base="https://api.pzero.studio/v1",
    api_key=pzero_api_key,
    context_window=128000,
    is_chat_model=True,
)

response = llm.complete("Hello from PZERO")
print(str(response))
```
