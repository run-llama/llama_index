# PopNow

[PopNow](https://popnow.ai/api) is a pay-as-you-go API that serves Claude, GPT, Gemini, DeepSeek, Kimi and other chat models through one OpenAI-compatible endpoint, billed per token in USD. Because the endpoint is OpenAI-compatible, it works with the `OpenAILike` LLM class; there is no separate package to install.

Create an API key in the [PopNow API console](https://popnow.ai/api/console/keys) and add funds to your API balance.

## Set up the LLM

PopNow model IDs follow the `<lab>/<model>/chat` format, for example `anthropic/claude-sonnet-5.5/chat` or `deepseek/v4-flash/chat`. The [chat API docs](https://popnow.ai/docs/chat) list every model with its context length and price.

```python
import os

from llama_index.llms.openai_like import OpenAILike

os.environ["POPNOW_API_KEY"] = "<your-popnow-api-key>"

llm = OpenAILike(
    model="google/gemini-3.7-flash/chat",
    api_base="https://api.popnow.ai/v1",
    api_key=os.environ["POPNOW_API_KEY"],
    context_window=1048576,
    is_chat_model=True,
    is_function_calling_model=True,
)
```

## Call `chat` with a ChatMessage list

```python
from llama_index.core.llms import ChatMessage

messages = [
    ChatMessage(role="system", content="You are a helpful assistant."),
    ChatMessage(role="user", content="Say this is a test."),
]
print(llm.chat(messages))
```

## Call `complete` with a prompt

```python
print(llm.complete("Who is Kaiming He"))
```

## Streaming

```python
for chunk in llm.stream_complete("Tell me what ResNet is"):
    print(chunk.delta, end="")
```

## Switch models

The same key works for every chat model, so you can compare models from different labs by changing the model ID.

```python
claude = OpenAILike(
    model="anthropic/claude-sonnet-5.5/chat",
    api_base="https://api.popnow.ai/v1",
    api_key=os.environ["POPNOW_API_KEY"],
    context_window=1000000,
    is_chat_model=True,
    is_function_calling_model=True,
)
print(claude.complete("Summarize the attention mechanism in one sentence."))
```
